# 02 – Data Type Handling

## 1. Introduction

Data Type Handling is an important step in Data Preprocessing. It involves identifying the data types of different columns and converting them into the correct formats.

Sometimes, numerical values are stored as strings, dates are stored as text, and categorical values are stored as numbers.

These incorrect data types can create problems during data analysis and Machine Learning.

Data Type Handling helps us identify and correct these problems so that the dataset can be processed properly.

**Objectives:**
- Identify different data types.
- Understand numerical and categorical data.
- Identify Boolean, Date/Time, and String data.
- Convert data types using Python.
- Handle invalid data conversions.
- Detect incorrect data types.

**Tools Used:** Python, Pandas, NumPy.



## 2. Identifying Data Types

Identifying data types means checking what type of information is stored in each column of a dataset.

Different columns may contain different types of information, such as numbers, text, dates, and Boolean values.

In Pandas, we can identify data types using:

- `df.dtypes` – Displays the data type of each column.
- `df.info()` – Displays column names, non-null values, and data types.
- `df.select_dtypes()` – Selects columns based on their data types.

**Example:**

A student dataset may contain:
- Age – Integer
- Marks – Float
- Name – String
- Gender – Categorical
- Attendance – Boolean
- Joining_Date – Date/Time

**Why is it important?**

Correctly identifying data types helps us select appropriate operations and preprocessing techniques for each column.



## 3. Different Types of Data

### 3.1 Numerical Data

Numerical data contains numbers that represent measurable or countable information.

There are two main types:

**Integer Data:**
- Contains whole numbers.
- Example: Age, Number of Students, Quantity.

**Float Data:**
- Contains decimal values.
- Example: Salary, Percentage, Temperature.

### 3.2 Categorical Data

Categorical data represents groups or categories.

Examples:
- Gender
- Department
- City
- Product Category

Categorical data can be divided into:

- Nominal: Categories without any order, such as Gender and City.
- Ordinal: Categories with a meaningful order, such as Low, Medium, and High.

### 3.3 Boolean Data

Boolean data represents only two possible values:

- True
- False

Example:
- Is_Passed
- Is_Active
- Is_Registered

### 3.4 Date/Time Data

Date/Time data represents dates and time information.

Examples:
- Date of Birth
- Joining Date
- Transaction Date

Pandas uses `datetime64` to store date and time information.

### 3.5 String Data

String data contains text or character information.

Examples:
- Student Name
- Address
- Email
- Phone Number

In Pandas, text columns may use `object` or `string` data types.



## 4. Converting Data Types

Data type conversion is the process of changing a column from one data type to another.

Sometimes, data is collected in an incorrect format. For example, numerical values may be stored as strings.

We need to convert these values into suitable data types before performing calculations or analysis.

Common examples:

1. String to Numeric
2. String to Date/Time
3. Numeric to Categorical
4. Integer to Float

**Why is conversion required?**

- To perform mathematical operations.
- To analyze dates correctly.
- To group categorical information.
- To reduce data processing errors.
- To prepare data for Machine Learning algorithms.



## 5. astype()

The `astype()` function in Pandas is used to convert a column or an entire DataFrame into a specified data type.

**Syntax:**

`df['column_name'].astype('data_type')`

**Examples:**

- Integer to Float
- Integer to String
- String to Category

**When to use it?**

Use `astype()` when the values are already compatible with the target data type.

**Limitation:**

If a column contains invalid values, direct conversion may generate an error.



## 6. to_numeric()

The `pd.to_numeric()` function is used to convert values into numerical data types.

It is especially useful when numerical values are stored as strings.

**Syntax:**

`pd.to_numeric(df['column_name'], errors='coerce')`

The `errors` parameter has three options:

- `raise`: Raises an error when conversion is not possible.
- `coerce`: Converts invalid values into NaN.
- `ignore`: Keeps the original values if conversion fails. This option is deprecated in newer Pandas versions.

**Why is it useful?**

It helps convert numerical strings into numbers and identify invalid numerical values.



## 7. to_datetime()

The `pd.to_datetime()` function is used to convert date values into a proper Date/Time format.

Sometimes dates are stored as strings, which makes date-based operations difficult.

**Syntax:**

`pd.to_datetime(df['column_name'], errors='coerce')`

**Examples:**

- String to Date/Time
- Identifying invalid dates
- Extracting year, month, and day

**Why is it useful?**

It helps perform date calculations, identify time-based patterns, and analyze information based on dates.

Invalid dates can be converted into NaT using `errors='coerce'`.



## 8. Handling Invalid Conversions

Invalid conversions occur when a value cannot be converted into the required data type.

**Examples:**

- Text such as "twenty" in an Age column.
- Text such as "unknown" in a Salary column.
- Invalid date such as "35-15-2026" in a Date column.

**Methods to handle invalid conversions:**

1. Identify invalid values.
2. Convert invalid values into NaN or NaT where appropriate.
3. Inspect the original values.
4. Decide whether to correct, impute, or remove the invalid records.

**Important:**

We should not blindly remove invalid values. We must first understand why the values are incorrect and choose an appropriate solution.



## 9. Detecting Incorrect Data Types

Detecting incorrect data types means identifying columns where the stored data type does not match the actual meaning of the information.

**Examples:**

- Age stored as String instead of Integer.
- Salary stored as String instead of Float.
- Joining_Date stored as String instead of Date/Time.
- Yes/No values stored as text instead of Boolean.

**Methods used:**

- `df.dtypes`
- `df.info()`
- `df.select_dtypes()`
- `pd.to_numeric()`
- `pd.to_datetime()`

**Why is it important?**

Incorrect data types can affect calculations, filtering, sorting, visualization, and Machine Learning operations.


```python

import pandas as pd
import numpy as np

print("Libraries imported successfully!")
```

```python

data = {
    'Student_ID': [101, 102, 103, 104, 105, 106],
    'Name': ['Ravi', 'Priya', 'Amit', 'Sneha', 'Kiran', 'Neha'],
    'Age': ['22', '23', 'twenty', '21', '24', '22'],
    'Study_Hours': ['4.5', '5.0', '3.5', 'six', '6.0', '4.0'],
    'Marks': [85, 90, 78, 88, 92, 80],
    'Gender': ['Male', 'Female', 'Male', 'female', 'Male', 'Female'],
    'Attendance': ['Yes', 'No', 'Yes', 'Yes', 'No', 'Yes'],
    'Joining_Date': [
        '10-01-2026',
        '15-02-2026',
        '20-03-2026',
        '35-15-2026',
        '25-04-2026',
        '30-05-2026'
    ],
    'Performance_Level': [1, 2, 3, 2, 1, 3]
}

df = pd.DataFrame(data)

display(df)
```

## Explanation:
- Age and Study_Hours are stored as strings.
- One Age value contains invalid text.
- One Study_Hours value contains invalid text.
- Gender contains inconsistent capitalization.
- Attendance contains Yes/No values.
- Joining_Date contains an invalid date.
- Performance_Level contains numerical category codes.

```python
print("Dataset Shape:", df.shape)

print("\nDataset Information:")
df.info()

print("\nColumn Data Types:")
print(df.dtypes)
```

## Explanation:
- shape displays the number of rows and columns.
- info() provides information about columns and non-null values.
- dtypes displays the data type of every column.

```python
print("Numerical Columns:")
print(df.select_dtypes(include=['number']).columns.tolist())

print("\nObject/String Columns:")
print(df.select_dtypes(include=['object', 'string']).columns.tolist())

print("\nBoolean Columns:")
print(df.select_dtypes(include=['bool']).columns.tolist())

print("\nDate/Time Columns:")
print(df.select_dtypes(include=['datetime']).columns.tolist())
```

```python

df['Age'] = pd.to_numeric(df['Age'], errors='coerce')

df['Study_Hours'] = pd.to_numeric(
    df['Study_Hours'], errors='coerce'
)

print("After Numeric Conversion:")
display(df[['Name', 'Age', 'Study_Hours']])

print("\nData Types:")
print(df[['Age', 'Study_Hours']].dtypes)
```

## Explanation:
- pd.to_numeric() converts compatible values into numbers.
- errors='coerce' converts invalid values into NaN.
- The invalid values twenty and six are now represented as NaN.

```python
df['Joining_Date'] = pd.to_datetime(
    df['Joining_Date'],
    format='%d-%m-%Y',
    errors='coerce'
)

print("After Date Conversion:")
display(df[['Name', 'Joining_Date']])

print("\nDate Data Type:")
print(df['Joining_Date'].dtype)
```

## Explanation:
- pd.to_datetime() converts strings into date values.
- format='%d-%m-%Y' specifies the date format.
- errors='coerce' converts invalid dates into NaT.

```python
df['Performance_Level'] = df['Performance_Level'].astype('category')

print("Performance Level:")
print(df['Performance_Level'])

print("\nData Type:")
print(df['Performance_Level'].dtype)
```

```python
df['Marks'] = df['Marks'].astype(float)

print("Marks After Conversion:")
print(df['Marks'])

print("\nData Type:")
print(df['Marks'].dtype)
```

## Explanation:
- astype(float) converts integer values into floating-point values.
- This is useful when decimal values may be required for calculations.

```python
df['Attendance'] = df['Attendance'].map({
    'Yes': True,
    'No': False
}).astype('boolean')

print("Attendance After Conversion:")
display(df[['Name', 'Attendance']])

print("\nData Type:")
print(df['Attendance'].dtype)
```

### Explanation:
- Yes is converted to True.
- No is converted to False.
- Pandas Boolean dtype is used to represent the values.

```python
df['Gender'] = df['Gender'].str.strip().str.title()

print("Gender Values After Standardization:")
print(df['Gender'].unique())
```

## Explanation:
- str.strip() removes unnecessary spaces.
- str.title() standardizes the capitalization.
- Values such as female and Female become consistent.

```python
print("Invalid Age Values:")
print(df.loc[df['Age'].isna(), 'Name'])

print("\nInvalid Study Hours Values:")
print(df.loc[df['Study_Hours'].isna(), 'Name'])

print("\nInvalid Joining Dates:")
print(df.loc[df['Joining_Date'].isna(), 'Name'])
```

```python
# Recreate original data for comparison

original_df = pd.DataFrame(data)

print("Original Invalid Age:")
print(original_df.loc[
    pd.to_numeric(original_df['Age'], errors='coerce').isna(),
    ['Name', 'Age']
])

print("\nOriginal Invalid Study Hours:")
print(original_df.loc[
    pd.to_numeric(original_df['Study_Hours'], errors='coerce').isna(),
    ['Name', 'Study_Hours']
])

print("\nOriginal Invalid Joining Date:")
print(original_df.loc[
    pd.to_datetime(
        original_df['Joining_Date'],
        format='%d-%m-%Y',
        errors='coerce'
    ).isna(),
    ['Name', 'Joining_Date']
])
```

## Explanation:

We preserved the original values so that we can identify exactly which records failed conversion.

We should investigate these records before deciding whether to correct them, impute them, or remove them.

```python
print("Final Column Data Types:")
print(df.dtypes)

print("\nNumerical Columns:")
print(df.select_dtypes(include=['number']).columns.tolist())

print("\nCategorical Columns:")
print(df.select_dtypes(include=['category']).columns.tolist())

print("\nBoolean Columns:")
print(df.select_dtypes(include=['bool', 'boolean']).columns.tolist())

print("\nDate/Time Columns:")
print(df.select_dtypes(include=['datetime']).columns.tolist())

print("\nString/Object Columns:")
print(df.select_dtypes(include=['object', 'string']).columns.tolist())
```

```python
print("Final Dataset:")
display(df)

print("\nDataset Shape:")
print(df.shape)

print("\nMissing Values:")
print(df.isnull().sum())

print("\nFinal Data Types:")
print(df.dtypes)
```
