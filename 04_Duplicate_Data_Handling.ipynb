# Duplicate Data Handling

## Introduction

Duplicate data occurs when the same record or information appears more than once in a dataset.

Duplicate records can affect data analysis and machine learning model performance.

They can increase the dataset size unnecessarily and may lead to incorrect results.

In this notebook, we will learn how to identify, analyze, and handle duplicate records using Python and Pandas.

## Objectives

- Understand duplicate records.
- Identify exact and partial duplicates.
- Detect duplicate records using Pandas.
- Remove duplicate records.
- Check duplicates based on selected columns.
- Handle duplicates using business rules.
- Compare the dataset before and after duplicate removal.

## 1. What is Duplicate Data?

Duplicate data means the same record or information is repeated in a dataset.

For example, if a student record is entered twice with the same student ID, name, age, and marks, it is considered a duplicate record.

### Why do duplicates occur?

- Repeated data entry.
- Combining multiple datasets.
- Data collection errors.
- Repeated form submissions.
- System or database errors.

### Problems caused by duplicate data

- Incorrect analysis.
- Increased dataset size.
- Biased machine learning results.
- Incorrect statistical calculations.
- Wasted storage and processing time.

```python
import pandas as pd
import numpy as np

print("Libraries imported successfully!")
```

## 2. Creating a Sample Dataset

We will create a student dataset containing duplicate records.

The dataset includes student ID, name, age, department, and marks.

We will use this dataset to practice detecting and removing duplicates.

```python
data = {
    "Student_ID": [101, 102, 103, 104, 105, 101, 106, 103, 107, 108],
    "Name": ["Asha", "Rahul", "Priya", "Kiran", "Sneha",
             "Asha", "Arun", "Priya", "Meena", "Ravi"],
    "Age": [20, 21, 22, 23, 21, 20, 24, 22, 20, 23],
    "Department": ["AI", "CS", "AI", "EC", "CS",
                   "AI", "AI", "AI", "EC", "CS"],
    "Marks": [85, 78, 90, 65, 88, 85, 76, 90, 82, 79]
}

df = pd.DataFrame(data)

display(df)
```

## 3. Understanding the Dataset

Before handling duplicates, we should inspect the dataset.

We can use:

- head() – Displays the first five rows.
- shape – Shows the number of rows and columns.
- info() – Displays column names and data types.
- describe() – Provides statistical information.

```python
print("First five records:")
display(df.head())

print("Dataset shape:", df.shape)

print("\nDataset information:")
df.info()

print("\nStatistical summary:")
display(df.describe())
```

## 4. Types of Duplicate Records

There are two important types of duplicates.

### 1. Exact Duplicates

Exact duplicates occur when all column values in two or more rows are the same.

Example:

| Student_ID | Name | Age | Department | Marks |
|---|---|---|---|---|
| 101 | Asha | 20 | AI | 85 |
| 101 | Asha | 20 | AI | 85 |

Both records contain exactly the same information.

### 2. Partial Duplicates

Partial duplicates occur when only selected column values are repeated, while other values may be different.

Example:

| Student_ID | Name | Age | Department | Marks |
|---|---|---|---|---|
| 101 | Asha | 20 | AI | 85 |
| 101 | Asha | 21 | AI | 88 |

The Student_ID and Name are the same, but Age and Marks are different.

Note: Partial duplicates need careful analysis because they may represent updated records rather than duplicate data.

## 5. Detecting Exact Duplicates

Pandas provides the duplicated() function to identify duplicate records.

By default, duplicated() checks all columns and marks repeated rows as True.

The first occurrence is marked False, and the following repeated occurrence is marked True.

```python
print("Duplicate records:")
print(df.duplicated())

print("\nTotal duplicate records:")
print(df.duplicated().sum())
```

```python
duplicate_rows = df[df.duplicated()]

print("Duplicate rows:")
display(duplicate_rows)
```

### Observation

The duplicated() function identifies rows that are repeated across all columns.

It helps us find exact duplicate records without modifying the original dataset.

## 6. Identifying All Duplicate Occurrences

By default, duplicated() uses keep="first".

We can use different keep options:

- keep="first" – Marks all repeated occurrences except the first as duplicates.
- keep="last" – Marks all repeated occurrences except the last as duplicates.
- keep=False – Marks every occurrence of a duplicate record as True.

```python
print("Keep first:")
display(df[df.duplicated(keep="first")])

print("Keep last:")
display(df[df.duplicated(keep="last")])

print("Mark all duplicate occurrences:")
display(df[df.duplicated(keep=False)])
```

## 7. Detecting Partial Duplicates

Sometimes records are repeated based on only a few important columns.

We can use the subset parameter in duplicated() to check selected columns.

For example, we can check whether Student_ID is repeated.

This is useful when Student_ID is expected to be unique.

```python
print("Duplicate Student IDs:")
display(df[df.duplicated(subset=["Student_ID"], keep=False)])

print("\nNumber of repeated Student IDs:")
print(df.duplicated(subset=["Student_ID"]).sum())
```

## 8. Checking Duplicates Using Multiple Columns

We can also check duplicates using more than one column.

For example, we can check whether the combination of Student_ID and Name is repeated.

This is useful when a combination of columns is expected to identify a unique record.

```python
duplicate_student_records = df[
    df.duplicated(subset=["Student_ID", "Name"], keep=False)
]

display(duplicate_student_records)
```

## 9. Removing Exact Duplicate Records

The drop_duplicates() function removes repeated records from a DataFrame.

By default:
- It checks all columns.
- It keeps the first occurrence.
- It removes the remaining duplicate occurrences.

We will create a copy of the original dataset before removing duplicates.

```python
clean_df = df.copy()

clean_df = clean_df.drop_duplicates()

print("Original dataset shape:", df.shape)
print("Dataset shape after removing duplicates:", clean_df.shape)

display(clean_df)
```

## 10. Keeping the Last Occurrence

Sometimes the latest record should be retained instead of the first record.

We can use keep="last" in drop_duplicates().

This keeps the last occurrence and removes the earlier repeated records.

```python
last_record_df = df.drop_duplicates(keep="last")

print("Dataset after keeping the last occurrence:")
display(last_record_df)
```

## 11. Removing Duplicates Based on Selected Columns

The subset parameter allows us to remove duplicates based on selected columns.

For example, if Student_ID is unique for every student, we can remove repeated Student_ID records.

We should choose which record to keep based on the business requirement.

```python
unique_students_df = df.drop_duplicates(
    subset=["Student_ID"],
    keep="first"
)

print("Dataset after removing repeated Student IDs:")
display(unique_students_df)

print("\nRemaining duplicate Student IDs:")
print(unique_students_df["Student_ID"].duplicated().sum())
```

## 12. Handling Partial Duplicates

Partial duplicates may have the same ID but different values in other columns.

We should not remove them automatically.

First, we need to understand why the records are repeated.

Let's create a separate dataset containing partial duplicates.

```python
partial_data = {
    "Student_ID": [201, 202, 203, 201, 204, 202],
    "Name": ["Anu", "Ravi", "Meena", "Anu", "Kiran", "Ravi"],
    "Marks": [75, 80, 90, 85, 70, 80],
    "Updated_Date": [
        "2026-11-01",
        "2026-11-02",
        "2026-11-03",
        "2026-11-05",
        "2026-11-04",
        "2026-11-02"
    ]
}

partial_df = pd.DataFrame(partial_data)

display(partial_df)
```

```python
partial_duplicates = partial_df[
    partial_df.duplicated(subset=["Student_ID"], keep=False)
]

print("Records with repeated Student IDs:")
display(partial_duplicates)
```

### Observation

Student_ID 201 appears more than once with different marks and update dates.

Student_ID 202 also appears more than once.

These records are not exact duplicates.

They may represent updated student information. We need a business rule to decide which record should be retained.

## 13. Handling Duplicates Using Business Rules

Business rules help us decide how to handle repeated records.

Examples:

- Keep the latest updated record.
- Keep the record with the highest data completeness.
- Keep the record from a trusted source.
- Keep the record with the most recent transaction date.

In our example, we will keep the latest updated record for each Student_ID.

We will first convert Updated_Date into a datetime data type, sort the records by date, and then keep the latest record.

```python
business_df = partial_df.copy()

business_df["Updated_Date"] = pd.to_datetime(
    business_df["Updated_Date"]
)

business_df = business_df.sort_values("Updated_Date")

latest_records = business_df.drop_duplicates(
    subset=["Student_ID"],
    keep="last"
)

print("Latest record for each student:")
display(latest_records.sort_values("Student_ID"))
```

### Observation

The latest updated record is retained for each Student_ID.

This approach is useful when repeated records represent updated information.

The business rule must be selected based on the actual requirements of the dataset.

## 14. Detecting Duplicates After Data Standardization

Sometimes records look different because of extra spaces or different letter cases.

For example:
- "Asha" and "asha"
- "Rahul" and " Rahul "

These values may represent the same person.

We can standardize text using strip() and title() before checking duplicates.

This helps us identify duplicate records more accurately.

```python
text_data = {
    "Student_ID": [301, 302, 303, 301],
    "Name": ["Asha", "Rahul", "Priya", " asha "],
    "Department": ["AI", "CS", "AI", "AI"]
}

text_df = pd.DataFrame(text_data)

print("Original dataset:")
display(text_df)

# Standardize text columns
text_df["Name"] = text_df["Name"].str.strip().str.title()
text_df["Department"] = text_df["Department"].str.strip().str.upper()

print("Dataset after standardization:")
display(text_df)

print("Duplicate records after standardization:")
display(text_df[text_df.duplicated()])
```

## 15. Counting Duplicate Records

We can calculate the number of duplicate records using duplicated().sum().

We can also calculate duplicate percentages.

Formula:

Duplicate Percentage = (Number of Duplicate Rows / Total Rows) * 100

```python
duplicate_count = df.duplicated().sum()

duplicate_percentage = (duplicate_count / len(df)) * 100

print("Total records:", len(df))
print("Duplicate records:", duplicate_count)
print("Duplicate percentage:", duplicate_percentage, "%")
```

## 16. Before-and-After Analysis

After removing duplicate records, we should compare the original dataset with the cleaned dataset.

We will check:

- Number of records before cleaning.
- Number of records after cleaning.
- Number of removed records.
- Number of remaining duplicates.

```python
before_count = len(df)
after_count = len(clean_df)

removed_count = before_count - after_count

print("Before duplicate removal:", before_count)
print("After duplicate removal:", after_count)
print("Removed duplicate records:", removed_count)

print("\nRemaining duplicates:")
print(clean_df.duplicated().sum())
```

## 17. Final Data Validation

After removing duplicates, we should validate the cleaned dataset.

We will check:
- Dataset shape.
- Duplicate record count.
- Unique student ID count.
- Column information.
- First few records.

```python
print("Final dataset shape:", clean_df.shape)

print("\nRemaining exact duplicates:")
print(clean_df.duplicated().sum())

print("\nUnique Student IDs:")
print(clean_df["Student_ID"].nunique())

print("\nDataset information:")
clean_df.info()

print("\nCleaned dataset:")
display(clean_df.head())
```

## 18. Saving the Cleaned Dataset

After cleaning and validating the dataset, we can save it as a CSV file.

The to_csv() function saves the DataFrame as a CSV file.

We use index=False to avoid adding an extra index column to the file.

```python
clean_df.to_csv("student_data_without_duplicates.csv", index=False)

print("Cleaned dataset saved successfully!")
```
