# 01 – Data Preprocessing Fundamentals

## 1. Introduction

Data preprocessing is an important step in Machine Learning. It is the process of converting raw data into clean, consistent, and meaningful data before using it to train a Machine Learning model.

Real-world datasets may contain missing values, duplicate records, incorrect data types, outliers, and inconsistent values.

These problems can affect the performance and accuracy of Machine Learning models.

Therefore, data preprocessing helps improve data quality and prepares the dataset for Machine Learning.

## 2. What is Data Preprocessing?

Data preprocessing is the process of cleaning, transforming, and organizing raw data into a suitable format for Machine Learning.

It involves:

- Handling missing values.
- Removing duplicate records.
- Correcting incorrect data types.
- Detecting and treating outliers.
- Encoding categorical variables.
- Scaling numerical features.
- Validating the dataset.

**Example:**

Suppose we have a student dataset containing:

- Missing student marks.
- Duplicate student records.
- Incorrect age values.
- Different formats of categorical values.

Before using this dataset for Machine Learning, we need to identify and correct these problems.

**Why is it required?**

It helps improve data quality, maintain consistency, reduce errors, and prepare reliable input for Machine Learning models.

## 3. Why is Data Preprocessing Required?

Data preprocessing is required because raw data is not always suitable for Machine Learning.

Main reasons:

1. **Missing Values:** Some records may have empty values.
2. **Duplicate Records:** The same information may appear multiple times.
3. **Incorrect Data Types:** Numerical values may be stored as strings.
4. **Outliers:** Some values may be extremely high or low.
5. **Inconsistent Data:** Categories may have different spellings or formats.
6. **Different Scales:** Numerical features may have different ranges.
7. **Categorical Data:** Text values need to be converted into numerical form for many ML algorithms.

**Real-world example:**

In a loan approval system, customer information may contain missing income values, duplicate applications, and incorrect age values.

Preprocessing helps identify and handle these issues before training a loan approval model.

## 4. Raw Data vs Clean Data

### Raw Data

Raw data is the original data collected from different sources before cleaning or preprocessing.

It may contain:
- Missing values.
- Duplicate records.
- Incorrect values.
- Inconsistent formats.
- Unnecessary columns.

### Clean Data

Clean data is data that has been checked, cleaned, and organized according to the requirements of the project.

It should have:
- Proper data types.
- Handled missing values.
- Appropriate duplicate handling.
- Consistent values.
- Valid records.

**Example:**

| Name | Age | Salary |
|---|---:|---:|
| Ravi | 25 | 30000 |
| Priya | NaN | 40000 |
| Ravi | 25 | 30000 |
| Amit | -5 | 35000 |

The above dataset contains a missing age, a duplicate record, and an invalid age.

After checking the records, we can handle these problems using suitable preprocessing techniques.

**Important:** Clean data does not mean deleting every unusual value. We must understand the data before making changes.

## 5. Data Quality

Data quality refers to how correct, complete, consistent, and reliable a dataset is.

Good data quality is important because Machine Learning models depend on the quality of the input data.

The main dimensions of data quality are:

1. Data Consistency
2. Data Completeness
3. Data Accuracy
4. Data Validity
5. Data Integrity

These dimensions help us identify whether the dataset is suitable for analysis and Machine Learning.

## 6. Data Consistency

Data consistency means that the same type of information should follow the same format and rules throughout the dataset.

**Example:**

A gender column may contain:

- Male
- male
- MALE

These values represent the same category but have different formats.

We can standardize them into one format, such as `Male`.

**Why is it important?**

It avoids treating the same information as different categories during data analysis and Machine Learning.

## 7. Data Completeness

Data completeness means that the required information is available in the dataset.

A dataset with many missing values may not provide enough information for reliable analysis.

**Example:**

| Name | Age | Marks |
|---|---:|---:|
| Ravi | 22 | 85 |
| Priya | NaN | 90 |
| Amit | 24 | NaN |

Here, Age and Marks contain missing values.

We need to analyze and handle these missing values using suitable techniques.

**Why is it important?**

Complete data helps Machine Learning models learn from sufficient information.

## 8. Data Accuracy

Data accuracy means that the values in a dataset correctly represent real-world information.

**Example:**

A student's actual age is 22, but the dataset contains 220.

This value is incorrect and should be investigated.

**Why is it important?**

Accurate data helps produce meaningful analysis and reliable Machine Learning predictions.

## 9. Data Validity

Data validity means that the data follows predefined rules, formats, and acceptable ranges.

**Example:**

- Age should not be negative.
- Marks should be between 0 and 100.
- Email addresses should follow a valid email format.
- Joining dates should follow an accepted date format.

**Why is it important?**

It helps identify invalid values and ensures that the dataset follows expected rules.

## 10. Data Integrity

Data integrity means maintaining the correctness and reliability of data throughout its lifecycle.

It ensures that data is not incorrectly modified, lost, or corrupted during storage, processing, or transformation.

**Example:**

In a student database, every student should have a unique Student ID.

If two different students have the same unique ID, it may create an integrity problem.

**Why is it important?**

It helps maintain trustworthy data and prevents incorrect relationships between records.

## 11. Data Leakage

Data leakage occurs when information that should not be available during model training is used to train the Machine Learning model.

It can make model performance appear better during training or evaluation than it will be in real-world situations.

**Example:**

Suppose we are building a model to predict whether a student will pass an examination.

If we use the final exam result as an input feature to predict whether the student will pass, the model already has the answer.

This is called target leakage.

**How to prevent it:**

- Split the dataset before fitting preprocessing techniques.
- Use only relevant input features.
- Fit scalers and imputers using training data only.
- Apply the fitted transformations to validation and test data.

**Why is it important?**

Preventing data leakage helps us evaluate the actual performance of a Machine Learning model.

## 12. Training Data, Validation Data and Test Data

A dataset is commonly divided into three parts before building a Machine Learning model.

### 1. Training Data

Training data is used to teach the Machine Learning model and learn patterns from the data.

### 2. Validation Data

Validation data is used to evaluate the model during development and tune its parameters.

### 3. Test Data

Test data is used to evaluate the final model on unseen data.

**Example:**

If we have 1,000 records, we can divide them as:

- Training data – 70% (700 records)
- Validation data – 15% (150 records)
- Test data – 15% (150 records)

**Why is it important?**

It helps us train, improve, and evaluate a model using separate datasets.

The test dataset should remain untouched until final evaluation.

## 13. Preprocessing Pipeline

A preprocessing pipeline is a sequence of steps used to clean and transform data before providing it to a Machine Learning model.

It helps organize preprocessing operations in a proper order.

**Common preprocessing steps:**

1. Load the dataset.
2. Inspect the dataset.
3. Identify data types.
4. Handle missing values.
5. Handle duplicate records.
6. Validate data.
7. Detect and treat outliers.
8. Encode categorical variables.
9. Scale numerical features.
10. Select relevant features.
11. Prepare data for Machine Learning.

**Why is it important?**

A preprocessing pipeline makes the workflow organized, repeatable, and easier to maintain.

It also helps prevent data leakage when preprocessing is correctly fitted on training data only.

### 1. Load Dataset

2. Inspect and Understand Data

3. Identify Data Quality Problems

4. Handle Missing Values and Duplicates

5. Validate Data and Treat Outliers

6. Encode Categorical Variables

7. Scale and Transform Features

8. Split Training, Validation and Test Data

9. Fit Preprocessing Pipeline on Training Data

10. Validate Final ML-Ready Data

## Python Implementation

```python
import pandas as pd
import numpy as np

print("Libraries imported successfully")
```

```python
data = {
    'Name': ['Ravi', 'Priya', 'Amit', 'Sneha', 'Ravi'],
    'Age': [22, np.nan, 24, 23, 22],
    'Gender': ['Male', 'Female', 'Male', 'female', 'Male'],
    'Marks': [85, 90, 78, np.nan, 85],
    'Salary': [30000, 40000, 35000, 45000, 30000]
}

df = pd.DataFrame(data)

df
```

```python
print("Dataset Shape:", df.shape)

print("\nDataset Information:")
df.info()

print("\nStatistical Summary:")
display(df.describe(include='all'))
```

```python
print("Missing Values:")
print(df.isnull().sum())
```

```python
print("Duplicate Records:")
print(df.duplicated().sum())
```

```python
df['Age'] = df['Age'].fillna(df['Age'].median())

df['Marks'] = df['Marks'].fillna(df['Marks'].median())

print("Missing values after handling:")
print(df.isnull().sum())
```

```python
df['Gender'] = df['Gender'].str.strip().str.title()

print(df['Gender'].unique())
```

```python
df = df.drop_duplicates()

print("Dataset after removing duplicates:")
display(df)
```

```python
print("Final Dataset Shape:", df.shape)

print("\nMissing Values:")
print(df.isnull().sum())

print("\nDuplicate Records:")
print(df.duplicated().sum())

print("\nFinal Dataset:")
display(df)
```
